# Amazn AI - Day 1
## Dataset Exploration & Preprocessing

```
import pandas as pd
import nltk
import matplotlib.pyplot as plt
import seaborn as sns
from nltk.tokenize import sent_tokenize
from sklearn.metrics.pairwise import cosine_similarity
nltk.download('punkt')
nltk.download('punkt_tab')
```

## Step 1: Load Dataset

```
df=pd.read_csv('../data/raw/amazon.csv')
df.head()
```

## Step 2: EDA

```
df.info()
df.describe(include='all')
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.columns.tolist())
```

## Step 3: Data Cleaning

```
df["discounted_price"]=df["discounted_price"].str.replace("₹","",regex=False).str.replace(",","",regex=False).astype(float)
df["actual_price"]=df["actual_price"].str.replace("₹","",regex=False).str.replace(",","",regex=False).astype(float)
df["discount_percentage"]=df["discount_percentage"].str.replace("%","",regex=False).astype(float)
df["rating"]=pd.to_numeric(df["rating"],errors="coerce")
df["category"]=df["category"].str.lower().str.split("|")
df.to_csv("../data/processed/amazon_cleaned.csv",index=False)
df.info()
```

## Step 4: Visualization

```
plt.figure(figsize=(10,5))
sns.histplot(df["rating"],bins=20)
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(df["actual_price"],bins=30,kde=True)
plt.show()
```

## Step 5: Chunking

```
def sentence_splitting(text,max_sentences=2):
    sentences=sent_tokenize(text)
    chunks=[]
    for i in range(0,len(sentences),max_sentences):
        chunks.append(" ".join(sentences[i:i+max_sentences]))
    return chunks

documents=[]
metadata=[]

for _,row in df.iterrows():
    category=" ".join(row["category"]) if isinstance(row["category"],list) else ""
    doc=f"""Product Name: {row['product_name']}

Category: {category}

Price: ₹{row['discounted_price']}

Rating: {row['rating']}

Description:
{row['about_product']}
"""
    documents.extend(sentence_splitting(doc))
    metadata.append({"product_name":row["product_name"],"price":row["discounted_price"],"rating":row["rating"]})

print("Documents:",len(documents))
documents[0]
```
